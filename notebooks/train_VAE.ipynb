```python
from scripts.train import train_model
from pathlib import Path
from src.utils.plots import loss_plot, train_val_plot
```

```python
model_type = "mlp"
seed = 42
batch_size = 64
image_size = 64
latent_dim = 64
epochs = 200
learning_rate = 0.001
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
data_dir = Path("/content/drive/MyDrive/Colab Notebooks/VAE/Cats")
checkpoint_path = Path("/content/drive/MyDrive/Colab Notebooks/VAE/checkpoints")
```

```python
import random
import numpy as np
import torch

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
torch.use_deterministic_algorithms(True)
```

```python
vae, train_losses, val_losses = train_model(
    model_type=model_type,
    seed=seed,
    batch_size=batch_size,
    image_size=image_size,
    latent_dim=latent_dim,
    epochs=epochs,
    learning_rate=learning_rate,
    data_dir=data_dir,
    checkpoint_path=checkpoint_path)
```
